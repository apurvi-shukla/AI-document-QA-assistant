```python
import numpy as np
import pandas as pd
from pypdf import PdfReader
from sentence_transformers import SentenceTransformer
from huggingface_hub import InferenceClient
```

```python
pdf_name = "ML-Notes.pdf"
reader = PdfReader(pdf_name)
print("Total pages:", len(reader.pages))
```

```python
first_page = reader.pages[0]
text = first_page.extract_text()
print(text)
```

```python
all_pages_text = []

for page_number, page in enumerate(reader.pages):
    page_text = page.extract_text()

    if page_text:
        all_pages_text.append({
            "page_number": page_number + 1,
            "text": page_text
        })

print("Total pages processed:", len(all_pages_text))
```

```python
print("Total extracted pages:", len(all_pages_text))
print("\nFirst page:")
print(all_pages_text[0]["text"])
```

```python
chunk_size = 1000
overlap = 200

print("Chunk size:", chunk_size)
print("Overlap:", overlap)
```

```python
chunks = []
for page in all_pages_text:

    text = page["text"]
    page_number = page["page_number"]

    start = 0
    while start < len(text):

        end = start + chunk_size
        chunk_text = text[start:end]
        chunks.append({
            "page_number": page_number,
            "text": chunk_text
        })
        start += chunk_size - overlap
print("Total chunks created:", len(chunks))
```

```python
print("Total chunks:", len(chunks))
print("\nFirst chunk:")
print(chunks[0]["text"])
print("\nSource page:", chunks[0]["page_number"])
```

```python
model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")
```

```python
sample_text = "Machine Learning allows computers to learn from data."
sample_embedding = model.encode(sample_text)
print(sample_embedding)
```

```python
print("Embedding shape:", sample_embedding.shape)
print("Number of dimensions:", len(sample_embedding))
```

```python
sentence1 = "Machine Learning helps computers learn from data."

sentence2 = "ML enables machines to learn automatically."

embedding1 = model.encode(sentence1)
embedding2 = model.encode(sentence2)

print("Embedding 1 shape:", embedding1.shape)
print("Embedding 2 shape:", embedding2.shape)
```

```python
similarity = np.dot(embedding1, embedding2) / (
    np.linalg.norm(embedding1) * np.linalg.norm(embedding2)
)
print("Cosine Similarity:", similarity)
```

```python
sentence3 = "The weather is sunny today."
embedding3 = model.encode(sentence3)
similarity_13 = np.dot(embedding1, embedding3) / (
    np.linalg.norm(embedding1) * np.linalg.norm(embedding3)
)
print("Similarity with related sentence:", similarity)
print("Similarity with unrelated sentence:", similarity_13)
```

```python
chunk_texts = [chunk["text"] for chunk in chunks]
chunk_embeddings = model.encode(
    chunk_texts,
    show_progress_bar=True
)
print("Chunk embeddings shape:", chunk_embeddings.shape)
```

```python
user_question = "What is Machine Learning?"
question_embedding = model.encode(user_question)
print("Question embedding shape:", question_embedding.shape)
```

```python
similarities = np.dot(
    chunk_embeddings,
    question_embedding
) / (
    np.linalg.norm(chunk_embeddings, axis=1) *
    np.linalg.norm(question_embedding)
)
print("Similarities shape:", similarities.shape)
```

```python
top_k = 5
top_indices = np.argsort(similarities)[::-1][:top_k]
print("Top 5 chunk indices:", top_indices)
```

```python
for rank, index in enumerate(top_indices, start=1):
    print(f"\nRank: {rank}")
    print("Chunk Index:", index)
    print("Page:", chunks[index]["page_number"])
    print("Similarity:", similarities[index])
    print("Text:", chunks[index]["text"][:500])
    print("-" * 60)
```

```python
retrieved_context = ""
for index in top_indices:
    retrieved_context += (
        f"\nSource Page: {chunks[index]['page_number']}\n"
        f"{chunks[index]['text']}\n"
    )
print(retrieved_context)
```

```python
prompt = f"""
You are an AI Document QA Assistant.

Answer the question using only the provided context.

If the answer is not available in the context,
clearly say that the document does not provide this information.

Context:
{retrieved_context}

Question:
{user_question}

Answer:
"""

print(prompt)
```

```python
def ask_question(user_question, top_k=5):

    # 1. Question embedding
    question_embedding = embedding_model.encode(user_question)

    # 2. Cosine similarity
    similarities = np.dot(
        chunk_embeddings,
        question_embedding
    ) / (
        np.linalg.norm(chunk_embeddings, axis=1) *
        np.linalg.norm(question_embedding)
    )

    # 3. Top K retrieval
    top_indices = np.argsort(similarities)[::-1][:top_k]

    # 4. Context building
    retrieved_context = ""

    for index in top_indices:
        retrieved_context += (
            f"\nSource Page: {chunks[index]['page_number']}\n"
            f"{chunks[index]['text']}\n"
        )

    # 5. Prompt creation
    prompt = f"""
    Answer using only the provided context.
    If information is unavailable, say so clearly.

    Context:
    {retrieved_context}

    Question:
    {user_question}
    """

    # 6. Gemini response
    interaction = gemini_client.interactions.create(
        model="gemini-3.6-flash",
        input=prompt,
        generation_config={
            "thinking_level": "low",
            "max_output_tokens": 500
        },
        timeout=30000,
        automatic_continuation=False
    )

    # 7. Source pages
    source_pages = sorted(
        set(chunks[i]["page_number"] for i in top_indices)
    )

    print("Question:", user_question)
    print("\nAI Answer:")
    print(interaction.output_text)
    print("\nSource PDF Pages:", source_pages)
```

```python
ask_question("What is Machine Learning?")
```
